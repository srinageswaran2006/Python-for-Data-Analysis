# **Seaborn**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import Markdown, display
```

## **Topic 1: Introduction to Seaborn**

```python
sales_df = pd.DataFrame({
     "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
     "Sales": [1200, 1500, 1700, 1600, 2000],
     "Profit": [200, 250, 300, 280, 350] })
sales_df
```

```python
sns.set_theme() #automatic spacing, grid, font(automatic clean version)
plt.plot(sales_df["Month"], sales_df["Sales"])
plt.title('Seaborn Theme Application')
plt.show()
```

```python
styles =["whitegrid", "darkgrid", "white", "ticks"] #type of styles to use
for style in styles:
  display(Markdown(f"## {style}"))
  sns.set_theme(style = style)
  plt.figure(figsize=(4,2))
  plt.plot(sales_df["Month"], sales_df["Sales"], marker="o")

  plt.title("Sales Trend")
  plt.xlabel("Month")
  plt.ylabel("Sales")
  plt.show()
```

## **Topic 2: Line Plot (sns.lineplot())**

```python
sns.set_theme(style="white")
sns.lineplot(data=sales_df, x="Month", y="Sales",marker="o") #need to explicitly mention data, x, y unlike plt
plt.show()
```

```python
trend_df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jan", "Feb", "Mar", "Apr", "May"],
    "Amount": [1200,1500,1700,1600,2000, 200,250,300,280,350],
    "Type": ["Sales"]*5 + ["Profit"]*5 })

sns.set_theme(style="white")
sns.lineplot(data=trend_df, x="Month", y="Amount",
             hue="Type", #automatically gives two lines else just avgs acc. to dataset
             marker="o",linewidth=2,
             palette={"Sales":"darkblue","Profit":"green"}) #customize line color
plt.title("Sales vs Profit using Hue")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.show()
```

## **Topic 3: Bar Plot (sns.barplot())**

```python
sales_df = pd.DataFrame({
    "Category": ["Technology", "Furniture", "Office Supplies"],
    "Sales": [850000, 740000, 720000]
})
```

**Major Pastelle to use:**
- "Set2"       # colorful, professional (1,2,3) -> (bright,soft,pastel)
- "pastel"     # soft colors
- "deep"       # Seaborn default style
- "Blues"      # blue shade
- "Blues_r"    # ranked sales charts
- "viridis"    # modern continuous look

```python
sns.set_theme(style="white")
sns.barplot(data=sales_df, x="Category", y="Sales", palette="Set2") #bar plot, pallete(refer the markdown)
plt.show()
```

```python
sns.set_theme(style="white")
plt.figure(figsize=(10,5))

sorted_df=sales_df.sort_values("Sales",ascending=False) #pre sort the DF before for sorting in Seaborn

sns.barplot(data=sorted_df,
            x="Sales",
            y="Category",
            palette="Set2")
plt.xticks(rotation=-315) #just to rotate the x axis
plt.show()
```

```python
orders_df = pd.DataFrame({
    "Category": ["Technology", "Technology",
                 "Furniture", "Furniture",
                 "Office Supplies", "Office Supplies"],
    "Sales": [1000, 1500, 800, 1200, 500, 700]
}) #sample datset for repeated categories
```

```python
sns.set_theme(style="white")
plt.figure(figsize=((8,5)))

# group by category and calculate total sales
# as_index=False keeps Category as a normal column
# sort_values(..., ascending=False) gives highest sales first

cat_sales=orders_df.groupby("Category", as_index=False)["Sales"].sum().sort_values(by="Sales",ascending=False)
ax= sns.barplot(data=cat_sales,
                x="Category",
                y="Sales",
                palette="Set2")

# enumerate() gives x-position (0,1,2) and the corresponding sales value
# ax.text(x, y, text) places value labels above each bar

for x, y in enumerate(cat_sales["Sales"]):
  ax.text(x, #xth position
          y +20, # yth position slightly above the bar
          str(y), #convert sales value to str
          ha="center") #central placement

plt.title("Total Sales by Category")
plt.show()
```

## **Topic 4: Count Plot (sns.countplot())**

```python
orders_df = pd.DataFrame({
    "Segment": [
        "Consumer", "Corporate", "Consumer",
        "Home Office", "Consumer", "Corporate",
        "Consumer", "Home Office", "Consumer"
    ]
})
```

```python
df = pd.DataFrame({
    "Category": [
        "Technology", "Furniture", "Technology",
        "Office Supplies", "Technology", "Furniture",
        "Office Supplies", "Technology"
    ]
})
```

```python
plt.figure(figsize=(8,5))
order=df["Category"].value_counts().index # get categories sorted by highest frequency

sns.countplot(data=df, x="Category", order=order, palette="Pastel1",edgecolor="black")
# add count labels on top of each bar
for container in ax.containers: #accessing container
  ax.bar_label(container) #labelling

plt.title("Category Frequency")
plt.xlabel("Category")
plt.ylabel("Frequency")
plt.show()
```

## **Topic 5: Histogram Plot (sns.histplot())**

```python
sales_values = [120, 450, 200, 1200, 350, 700, 900, 150, 300, 400,
                250, 600, 800, 1000, 110, 220, 330, 550, 720, 950]

sales_df = pd.DataFrame({
    "Sales": sales_values
})
```

```python
sns.histplot(data=sales_df, x="Sales") #Histogram Plot
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=sales_df, x='Sales', bins=8,
             kde=True, #KDE gives smooth line est. of data distribution
             color="purple", edgecolor="black",linewidth=1.5,)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.grid(axis="y")
plt.show()
```

```python
#Different Bins

plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
sns.histplot(data=sales_df, x="Sales", kde=True, bins=5, color="teal", edgecolor="black")
plt.title("5 Bins")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.subplot(1,2,2)
sns.histplot(data=sales_df, x='Sales', kde=True, bins=12, color="orange", edgecolor="black")
plt.title("12 Bins")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

## **Topic 6: Box Plot (sns.boxplot())**

```python
sales_data = [120, 150, 180, 200, 220, 250, 270, 300, 320,
              350, 400, 450, 500, 550, 600, 700, 800, 1200, 2500]

sales_df = pd.DataFrame({
    "Sales": sales_data
})
sales_df.describe()
```

```python
plt.figure(figsize=(14,8))
plt.subplot(1,2,1)
sns.boxplot(data=sales_df, y="Sales") #vertical box plot
plt.title("Vertical Sales Box Plot")

plt.subplot(1,2,2)
sns.boxplot(data=sales_df, x="Sales") #horizontal box plot
plt.title("Horizontal Sales Box Plot")

plt.tight_layout()
plt.show()
```

```python
compare_df = pd.DataFrame({
    "Category": ["Furniture"]*8 + ["Technology"]*8,

    "Sales": [120,150,180,220,300,450,700,1200,
              200,250,320,400,500,600,800,2500]
})
```

```python
plt.figure(figsize=(8,8))
sns.boxplot(data=compare_df,x="Category", y="Sales", palette="Pastel2",linewidth=2)
plt.title("Sales Comparison")
plt.xlabel("Category")
plt.ylabel("Sales")
```

## **Topic 7: Violin Plot (sns.violinplot())**

```python
compare_df = pd.DataFrame({
    "Category": ["Furniture"]*8 + ["Technology"]*8,

    "Sales": [120,150,180,220,300,450,700,1200,
              200,250,320,400,500,600,800,2500]
})
```

```python
plt.figure(figsize=(10,10))
sns.violinplot(data=compare_df, x="Category", y="Sales", #violin plot
               inner="quartile", #to get quartiles inside the graph
               palette="Pastel2",
               linewidth=2)
plt.yticks(range(-1000,4001,500)) #to set range and step as 500
plt.title("Sales Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
#comparing violin plot and box plot
plt.figure(figsize=(10,10))

plt.subplot(1,2,1)
sns.violinplot(data=compare_df,x="Category", y="Sales", inner="quartile", palette="Pastel2",linewidth=2,
               cut=0) # to cut exactly and min and max values
plt.yticks(range(-1000,4001,500))
plt.title("Sales Distribution by Category(Violin Plot)")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.subplot(1,2,2)
sns.boxplot(data=compare_df,x="Category", y="Sales", palette="Pastel2",linewidth=2)
plt.title("Sales Distribution by Category(Box Plot)")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()
```

## **Topic 8: Scatter Plot (sns.scatterplot())**

```python
sales_profit_df = pd.DataFrame({
    "Sales": [100, 200, 300, 400, 500, 600, 700, 800],
    "Profit": [10, 25, 40, 55, 65, 80, 95, 110],
    "Category": [
        "Furniture", "Furniture", "Furniture",
        "Technology", "Technology", "Technology",
        "Office Supplies", "Office Supplies"
    ]
})
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=sales_profit_df, x="Sales", y="Profit")
plt.title("Sales vs Profit")
plt.show()
```

```python
sales_profit_df["Quantity"] = [1,2,2,3,4,3,5,4]
plt.figure(figsize=(8,5))

sns.scatterplot(data=sales_profit_df, x="Sales",y="Profit",
                hue="Category", # color by category

                palette={       # custom colors
                    "Furniture": "red",
                    "Technology": "blue",
                    "Office Supplies": "green"},

                size="Quantity",  # point size from Quantity

                sizes=(50,300),   # min and max point size

                style="Category", # different marker shapes

                # s=120, # fixed base marker size (remove if using size)
                alpha=0.7  # transparency
                )
plt.title("Sales vs Profit with Category")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=sales_profit_df, x="Sales", y="Profit",
                hue="Category",
                size="Quantity",
                sizes=(80,250),
                alpha=0.8,
                edgecolor="black")
plt.title("Sales vs Profit with Analysis(with Category)")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.grid(True, alpha=0.3)
plt.show()
```

## **Topic 9: Regression Plot (sns.regplot())**

```python
sales_profit_df = pd.DataFrame({
    "Sales": [100, 200, 300, 400, 500, 600, 700, 800],
    "Profit": [10, 25, 40, 55, 65, 80, 95, 110]
})
```

```python
sns.regplot(data=sales_profit_df, x="Sales", y="Profit",ci=95) #Regression Plot
plt.title("Sales vs Profit with Regression")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.regplot(data=sales_profit_df, x="Sales", y="Profit",
                scatter_kws={      # customize scatter points
                    "s":100,       #size
                    "color":"purple",
                    "edgecolor":"black",
                    "alpha":0.8},
                line_kws={         # customize regression line
                    "color":"darkred",
                    "linewidth":2,
                    "linestyle" : "--"},
                ci=None)           # remove confidence interval band
plt.title("Sales vs Profit Regression Analysis")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.grid(True, alpha=0.3)
plt.show()
```

## **Topic 10: Heatmap (sns.heatmap())**

```python
sales_df = pd.DataFrame({
    "Sales": [100, 200, 300, 400, 500, 600, 700, 800],
    "Profit": [10, 25, 40, 55, 65, 80, 95, 110],
    "Quantity": [1, 2, 2, 3, 4, 3, 5, 4],
    "Discount": [0.10, 0.05, 0.15, 0.20, 0.10, 0.25, 0.05, 0.30]
})
```

```python
corr_mat = sales_df.corr(numeric_only=True) #first create a correlation matrix(with numeric data)
corr_mat
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(corr_mat,annot=True) #Heatmap
plt.title("Correlation Heatmap")
plt.show()
```

```python
corr_map_col=["coolwarm","Blues","Greens","viridis","magma","YlGnBu"] #Heatmap Color Palette [Rec: coolwarm]

for color in corr_map_col:
  plt.figure(figsize=(8,5))
  sns.heatmap(corr_mat, annot=True, cmap=color) #cmap to assign palette
  plt.title(f"Correlation Heatmap ({color})")
  plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.heatmap(data=corr_mat,annot=True,cmap="coolwarm",
            fmt=".2f", #format values to 2 decimals
            linewidths=1, #create line between cells(with width)
            linecolor="white") #line color
plt.title("Correlation Heatmap")
plt.show()
```

## **Topic 11: Pair Plot (sns.pairplot())**

```python
sales_df = pd.DataFrame({
    "Sales": [100, 200, 300, 400, 500, 600, 700, 800],
    "Profit": [10, 25, 40, 55, 65, 80, 95, 110],
    "Quantity": [1, 2, 2, 3, 4, 3, 5, 4],
    "Discount": [0.10, 0.05, 0.15, 0.20, 0.10, 0.25, 0.05, 0.30]
})
```

```python
sns.pairplot(sales_df) #Pair Plot
plt.show()
```

```python
sales_df["Category"] = [
    "Furniture", "Furniture",
    "Technology", "Technology",
    "Technology", "Office Supplies",
    "Office Supplies", "Office Supplies"
]
```

```python
sns.pairplot(sales_df,
             hue="Category",
             vars=["Sales","Profit","Quantity"], #tocombine only particular columns
             diag_kind="kde", #to create smooth density curves [default:hist]
             height=3, # Height of sub plots
             palette={"Furniture":"red","Technology":"green","Office Supplies":"blue"}, #color customisation
             plot_kws={"s":75}) #plot point size
plt.show()
```

## **When to Use Which Plot?**

- Compare category averages : **barplot()**

- Count occurrences : **countplot()**

- Understand distribution : **histplot()**

- Detect outliers : **boxplot()**

- See distribution shape : **violinplot()**

- Analyze relationships : **scatterplot()**

- Add trend line : **regplot()**

- Analyze correlations : **heatmap()**

- Explore many variables quickly : **pairplot()**
