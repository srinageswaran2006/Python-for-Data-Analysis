## **Phase 6: Data Visualization (Matplotlib)**

```python
import matplotlib.pyplot as plt
```

### 6.1 Monthly Sales Trend

```python
monthly_sales= df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()
monthly_sales.index=monthly_sales.index.to_timestamp().strftime("%b-%Y") #converting period to timestamp(datetime)

plt.figure(figsize=(12,4))
plt.plot(monthly_sales.index,monthly_sales.values, marker="o",color="green",mec="black",mew=1.5,mfc="white")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(monthly_sales.index[::2], rotation=45) #to skip alternative month name for space
plt.grid(True)
plt.show()
```

### **Observation**
- Sales fluctuate considerably across the **2014–2017** period, indicating noticeable seasonal variation.

- The highest monthly sales occur in late 2017, where sales exceed **115,000**, making it the strongest-performing month in the dataset.

- Sales repeatedly increase during the **September–November period**, suggesting a recurring year-end sales surge.

- The overall trend from 2014 to 2017 is **upward**, with both peak sales and average monthly sales increasing over time.

### 6.2 Monthly Sales and Profit Comparison

```python
monthly_sp=df.groupby(df["Order Date"].dt.to_period("M"))[["Sales","Profit"]].sum()
monthly_sp.index=monthly_sp.index.to_timestamp().strftime("%b-%Y") #converting period to timestamp(datetime)

plt.figure(figsize=(12,4))
plt.plot(monthly_sp.index,monthly_sp["Sales"] ,label="Sales", marker="o",linewidth=2, color="blue",mec="black",mew=1.5,mfc="white")
plt.plot(monthly_sp.index,monthly_sp["Profit"],label="Profit", marker="o",linewidth=2, color="green",mec="black",mew=1.5,mfc="white")
plt.title("Monthly Sales and Profit Comparison")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.xticks(monthly_sp.index[::2],rotation=45) #to skip alternative month name for space
plt.legend(loc="upper left")
plt.tight_layout()
plt.grid(True)
plt.show()
```

### **Observation**
- Sales and Profit generally move in the **same direction**, showing that higher sales usually contribute to higher profit.

- Profit values are much **smaller in magnitude** than sales, indicating that revenue growth does not translate proportionally into profit growth.

- Some months show **strong sales but relatively lower profit**, suggesting the possible impact of discounts, shipping costs, or lower-margin products.

- The strongest combined sales and profit performance occurs during the **second half of 2017**.

### 6.5 Sales Distribution

### **Observation**

### 6.6 Sales vs Profit Relationship

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Sales"],
            df["Profit"],
            s=25,
            alpha=0.5,
            color="steelblue")

# trend line
m, b = np.polyfit(df["Sales"], df["Profit"], 1) # y=mx+b
plt.plot(df["Sales"],
         m * df["Sales"] + b,
         color="red",
         linewidth=2)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.xlim(0, 10000) #x axis limit
plt.ylim(-4000, 6000) #y axis limit

plt.grid(True, alpha=0.3) #alpha is transparency(0-1[trans-opa])
plt.tight_layout()
plt.show()
```

### **Observation**
- The scatter plot shows a **moderate positive relationship** between Sales and Profit, supported by the upward-sloping trend line.

- Most transactions are concentrated in the **0–3,000 sales range** with profits clustered near zero.

- Several orders with sales between **1,000 and 4,000** generate **negative profit**, indicating that higher sales do not always guarantee profitability.

- A few higher-sales transactions produce **substantially higher profits**, but the spread of points shows considerable variability in profit outcomes across similar sales levels.
