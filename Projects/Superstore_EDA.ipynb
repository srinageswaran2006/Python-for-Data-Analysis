## **Phase 3 – Univariate Analysis**

### **Part A - Numerical Features**

```python
num_cols=["Sales","Profit","Quantity","Discount"]
```

```python
#mean, median, mode, kurtosis, skewness

df[num_cols].agg(["mean","median",
                  lambda x:
                  x.mode().iloc[0] if not x.mode().empty else None,"kurt","skew"]
                 ).rename(index={"<lambda>":"mode"}
        ).T
```

```python
df[num_cols].var()
```

```python
df[num_cols].describe().T # table format change
```

### **Observations**

- Sales and Profit have a noticeable difference between their mean and median, suggesting that a few high-value transactions increase the average.
- Quantity and Discount have mean and median values that are quite close, indicating a more balanced distribution.
- **Sales** is the most positively skewed feature, followed by **Profit**, while **Quantity** and **Discount** are comparatively less skewed.
- **Sales** and **Profit** contain several high-value outliers, whereas **Quantity** and **Discount** have relatively fewer extreme values.
- **Sales** and **Profit** show a much wider range of values compared to **Quantity** and **Discount**, indicating greater variation.
- No major anomalies or invalid values were found in the numerical columns.

- The **West region** has the highest number of transactions, although the distribution across all four regions is fairly balanced.
- **Office Supplies** is the most purchased category, contributing around **60%** of all transactions.
- **Binders** are the most frequently purchased sub-category, whereas **Copiers** have the fewest transactions.
- The categorical columns contain consistent values with no obvious inconsistencies.

## **Phase 4 – Bivariate Analysis**

```python
df[num_cols].corr()
```

```python
for x in cat_cols:
  print(f"\n{'='*50}")
  display(Markdown(f"###**{x}**"))
  grouped=grouped = df.groupby(x).agg({
    "Sales": ["sum", "mean"],
    "Profit": ["sum", "mean"],
    "Quantity": "sum",
    "Discount": "mean"
})
  grouped.columns=["_".join(map(str,col))for col in grouped.columns]
  grouped=grouped.sort_values("Sales_sum",ascending=False)
  display(grouped)
```

```python
piv_reg=pd.pivot_table(df,index="Region",values=["Sales","Profit","Discount"],columns="Category",aggfunc="mean",sort=False)
piv_reg.columns=("_".join(map(str,col)) for col in piv_reg.columns)
piv_reg.T
```

### **Observations**

- Sales and Profit show a moderate positive correlation (**r = 0.48**), while Discount has a weak negative correlation with Profit (**r = -0.22**), indicating that higher discounts tend to reduce profitability.
- **Standard Class** contributes the highest total sales, whereas **First Class** generates the highest average profit per transaction.
- The **Consumer** segment contributes the highest total sales, while the **Home Office** segment records the highest average profit per order.
- The **West** region generates the highest total sales and profit, whereas the **Central** region has the lowest average profit and the highest average discount.
- **Technology** is the best-performing category in terms of average sales and profit, while **Furniture** generates high sales but comparatively lower profit.
- **Copiers** generate the highest average sales and profit per transaction despite having a low sales volume. In contrast, **Tables, Bookcases, and Supplies** are the only loss-making sub-categories.
- **Furniture** in the **Central** region records a negative average profit along with the highest average discount, suggesting that heavy discounting may be reducing profitability.

### Observations

- Sales and profit generally increase over the years, indicating steady business growth between **2014** and **2017**.
- Higher discount levels (especially **30% and above**) are associated with negative average profits. **Binders**, **Machines**, and **Tables** receive the highest average discounts, suggesting that aggressive discounting may be reducing profitability.
