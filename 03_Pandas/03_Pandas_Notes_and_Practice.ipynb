## **Pivot Tables**

### **Dataset**

```python
sales_data = {
    "Region": ["North","North","North","South","South","South","East","East","West","West"],
    "Category": ["Electronics","Furniture","Electronics","Furniture","Electronics","Furniture","Electronics","Furniture","Electronics","Furniture"],
    "Salesperson": ["Alice","Bob","Alice","Charlie","David","Charlie","Emma","Emma","Frank","Frank"],
    "Gender": ["F","M","F","M","M","M","F","F","M","M"],
    "Sales": [5000,3000,7000,4000,6000,3500,8000,4500,9000,5000],
    "Profit": [1000,500,1800,700,1200,600,2000,900,2500,1000]
}

sales_df = pd.DataFrame(sales_data)

sales_df
```

### **Code**

```python
pd.pivot_table(sales_df,index="Region",values="Sales",aggfunc="mean")
```

```python
pd.pivot_table(sales_df,index="Category",values="Profit",aggfunc="sum")
```

```python
pd.pivot_table(sales_df,
               index="Region"
               ,columns="Category"
               ,values="Sales"
               ,aggfunc="sum",
               sort = False  # to make table not to sort
)
```

```python
pd.pivot_table(sales_df,
               index=["Region","Category"],
               columns="Gender",
               values="Sales",
               aggfunc="mean",
               dropna=False,) # used when an entire NaN column is needed(actually deletes automatically if so)
```

```python
pivot=pd.pivot_table(sales_df,index="Region",values=["Sales","Profit"],aggfunc="mean")

pivot.rename(columns={"Profit":"Avg Profit","Sales":"Avg Sales"})
```

```python
pd.pivot_table(sales_df,
               index="Region",
               values="Sales",
               aggfunc=["mean","sum","count","max"],
               margins=True, margins_name="Grand Total") #grand total[for mean it gives overall avg]
```

```python
pivot1=pd.pivot_table(sales_df,
               index=["Region","Category"],
               columns="Gender",values=["Sales","Profit"],
               aggfunc=["mean","sum"],
               fill_value=0,  #filling Nan with 0
               )
pivot1.columns=["_".join(map(str,col))    # used for Flattening MultiIndex columns(rename)
                for col in pivot1.columns
]
display(pivot1)
```

```python
# Different aggregation for each value column
pd.pivot_table(
    sales_df,
    index="Region",
    values=["Sales", "Profit"],
    aggfunc={
        "Sales": "mean",
        "Profit": "sum"
    }
)
```

## **map(), apply(), and lambda**

### map()

### **Dataset**

```python
student_data = {
    "Student":["Ram","John","Sara","Emma","Alex"],
    "Gender":["M","M","F","F","M"],
    "Result":["P","F","P","P","F"]
}

student_df = pd.DataFrame(student_data)
student_df2 = student_df.copy() #because once it maps then old one is deleted
```

### **Code**

```python
student_df2["Gender"]=student_df2["Gender"].map({"M":"Male","F":"Female"})
```

```python
student_df2["Result"]=student_df2["Result"].map({"P":"Pass","F":"Fail"})
```

```python
student_df2["Scholarship"]=student_df2["Result"].map({"Pass":"Eligible","Fail":"Not Eligible"})
student_df2
```

### apply()

### **Dataset**

```python
# apply()

student_data = {
    "Student":["Ram","John","Sara","Emma","Alex"],
    "Marks":[95,72,35,88,40]
}

student_df = pd.DataFrame(student_data)
```

### **Code**

```python
def result(Marks):
  if Marks >=40:
    return "Pass"
  else:
    return "Fail"

student_df["Result"]=student_df["Marks"].apply(result)
student_df
```

```python
def grade(Marks):
  if Marks>=90:
    return "A"
  elif Marks >=75 and Marks<90:
    return "B"
  elif Marks>=40 and Marks<75:
    return "C"
  else:
    return "D"

student_df["Grade"]=student_df["Marks"].apply(grade)
student_df
```

```python
def b_mark(Marks):
  if Marks>=95: #capped to 100
    return 100
  else:
    return Marks+5

student_df["Bonus Marks"]=student_df["Marks"].apply(b_mark)
student_df
```

### lambda

### **Code**

```python
x=16

print("Value: ",(lambda x : x*2)(x))

print("Odd/Even: ",(lambda x:"Even" if x%2==0 else "Odd")(x))

print("Adult/Child: ",(lambda x: "Adult" if x>=18 else "Child")(x))
```

### apply() + lambda

### **Dataset**

```python
student_data = {
    "Student":["Ram","John","Sara","Emma"],
    "Marks":[95,85,38,72],
    "Attendance":[92,60,95,80]
}

student_df = pd.DataFrame(student_data)
```

### **Code**

```python
student_df["Double Marks"]=student_df["Marks"].apply(lambda x:x*2) #just for practice purpose
student_df
```

```python
student_df["Result"]=student_df["Marks"].apply(lambda x: "Pass" if x >=40 else "Fail")
student_df
```

```python
student_df["Bonus Marks"]=student_df["Marks"].apply(lambda x: 100 if x>=95 else x+5)
student_df
```

```python
student_df["Final Result"]=student_df.apply(lambda row:
                                            "Pass" if row["Marks"]>=40 and row["Attendance"]>=75
                                            else "Fail",
                                            axis=1 #row wise(0 for column)
                                            )
student_df
```

```python
student_df["Grade"]=student_df.apply(lambda row:
                                     "F" if row["Marks"]<40
                                     else
                                     "A" if row["Marks"]>=90 and  row["Attendance"]>=90
                                     else
                                     "B" if row["Marks"]>=75
                                     else "C",
                                     axis=1
                                     )
student_df
```
