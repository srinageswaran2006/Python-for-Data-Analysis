## **GroupBy**

```python
import pandas as pd

employee_data = {
    "Employee": ["Alice", "John", "Bob", "Charlie", "David", "Emma", "Kevin", "Sophia"],
    "Department": ["Finance", "Finance", "HR", "IT", "Finance", "HR", "IT", "Finance"],
    "Gender": ["Female", "Male", "Male", "Male", "Male", "Female", "Male", "Female"],
    "Salary": [60000, 50000, 50000, 70000, 70000, 65000, 75000, 80000]
}

emp_df = pd.DataFrame(employee_data)
```

```python
emp_df.groupby("Department")["Salary"].mean()
```

```python
emp_df.groupby("Department")["Salary"].sum()
```

```python
emp_df.groupby("Department")["Employee"].count()
```

```python
emp_df.groupby("Department")["Salary"].agg(["mean","sum","min","max","count"])
```

```python
emp_df.groupby(["Department","Gender"])["Salary"].agg("mean")
```

## **Pivot Tables**

```python
sales_data = {
    "Region": ["North","North","North","South","South","South","East","East","West","West"],
    "Category": ["Electronics","Furniture","Electronics","Furniture","Electronics","Furniture","Electronics","Furniture","Electronics","Furniture"],
    "Salesperson": ["Alice","Bob","Alice","Charlie","David","Charlie","Emma","Emma","Frank","Frank"],
    "Gender": ["F","M","F","M","M","M","F","F","M","M"],
    "Sales": [5000,3000,7000,4000,6000,3500,8000,4500,9000,5000],
    "Profit": [1000,500,1800,700,1200,600,2000,900,2500,1000]
}

sales_df = pd.DataFrame(sales_data)

sales_df
```

```python
pd.pivot_table(sales_df,index="Region",values="Sales",aggfunc="mean")
```

```python
pd.pivot_table(sales_df,index="Category",values="Profit",aggfunc="sum")
```

```python
pd.pivot_table(sales_df,
               index="Region"
               ,columns="Category"
               ,values="Sales"
               ,aggfunc="sum",
               sort = False  # to make table not to sort
)
```

```python
pd.pivot_table(sales_df,
               index=["Region","Category"],
               columns="Gender",
               values="Sales",
               aggfunc="mean",
               dropna=False,) # used when an entire NaN column is needed(actually deletes automatically if so)
```

```python
pivot=pd.pivot_table(sales_df,index="Region",values=["Sales","Profit"],aggfunc="mean")

pivot.rename(columns={"Profit":"Avg Profit","Sales":"Avg Sales"})
```

```python
pd.pivot_table(sales_df,
               index="Region",
               values="Sales",
               aggfunc=["mean","sum","count","max"],
               margins=True, margins_name="Grand Total") #grand total[for mean it gives overall avg]
```

```python
pivot1=pd.pivot_table(sales_df,
               index=["Region","Category"],
               columns="Gender",values=["Sales","Profit"],
               aggfunc=["mean","sum"],
               fill_value=0,  #filling Nan with 0
               )
pivot1.columns=["_".join(map(str,col))    # used for Flattening MultiIndex columns(rename)
                for col in pivot1.columns
]
display(pivot1)
```

```python
# Different aggregation for each value column
pd.pivot_table(
    sales_df,
    index="Region",
    values=["Sales", "Profit"],
    aggfunc={
        "Sales": "mean",
        "Profit": "sum"
    }
)
```
