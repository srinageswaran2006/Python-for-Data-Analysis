# **Matplotlib**

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
profit_df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Profit": [200, 250, 300, 280, 350]
 })
```

```python
plt.plot(profit_df["Month"],profit_df["Profit"])
```

```python
plt.figure(figsize=(10,5)) # Set graph size
plt.plot(profit_df["Month"],profit_df["Profit"],marker="s",color="red") # Plot line with marker and color
plt.xlabel("Month") # Set x-axis label
plt.ylabel("Profit") # Set y-axis label
plt.title("Monthly Profit Trend") # Set graph title
plt.grid(True) # Add gridlines
# plt.show() # Display the plot
```

```python
plt.figure(figsize=(12,4)) # Set graph size
plt.plot(profit_df["Month"],profit_df["Profit"],marker="^",color="blue",linestyle="--") # Plot line with marker, color, and dashed style
plt.title("Monthly Profit Trend") # Set graph title
plt.xlabel("Month") # Set x-axis label
plt.ylabel("Profit") # Set y-axis label
plt.grid(True) # Add gridlines
plt.show()
```

```python
category_df = pd.DataFrame({
    "Category": ["Furniture", "Technology", "Office Supplies"],
    "Sales": [742000, 836000, 719000]
})

category_df
```

```python
plt.bar(category_df["Category"],category_df["Sales"])
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.bar(category_df["Category"],category_df["Sales"],color="green") # Plot bars with color
plt.title("Category Sales Comparison")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.grid(axis="y") # Add gridlines only on y-axis
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.bar(category_df["Category"],category_df["Sales"],color="purple",edgecolor="black",linewidth=2) # Plot bars with color, edge color, and line width
plt.title("Category Sales Comparison")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
plt.barh(category_df["Category"],category_df["Sales"])
```

```python
plt.figure(figsize=(8,5))
plt.barh(category_df["Category"],category_df["Sales"],color="orange") # Plot horizontal bars with color
plt.title("Category Sales Comparison")
plt.xlabel("Sales")
plt.ylabel("Category")
plt.grid(axis="x") # Add gridlines only on x-axis
plt.show()
```

```python
plt.figure(figsize=(9,5))
plt.barh(category_df["Category"],category_df["Sales"],color="purple",edgecolor="black",linewidth=2,label="Sales") # Plot horizontal bars with style and label
plt.title("Category Sales Comparison")
plt.xlabel("Sales")
plt.ylabel("Category")
plt.grid(axis="x")
plt.legend() # Display legend
plt.show()
```

```python
compare_df = pd.DataFrame({
    "Sales": [120, 150, 180, 200, 220, 250, 270, 300, 320,
                                      350, 400, 450, 500, 550, 600, 700, 800, 1200, 2500],

                           "Profit": [10, 12, 15, 18, 20, 22, 25, 27, 30, 32, 35, 40,
                                      42, 45, 50, 55, 60, 120, 300] })
# Display the DataFrame (optional)
```

```python
plt.figure(figsize=(5,5))
box= plt.boxplot([compare_df["Sales"],
                  compare_df["Profit"]],
                 tick_labels=["Sales","Profit"], # Use tick_labels as 'labels' is deprecated
                 patch_artist=True)  # Enable filling boxes with color
box["boxes"][0].set_facecolor("red") # Set color for first box
box["boxes"][1].set_facecolor("blue") # Set color for second box
plt.title("Sales and Profit Comparison")
plt.ylabel("Value")
plt.show()
```

```python
sales_df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
                         "Sales": [1200, 1500, 1700, 1600, 2000],
                          "Profit": [200, 250, 300, 280, 350] })
sales_df
```

```python
plt.figure(figsize=(8,4))
plt.plot(sales_df["Month"],sales_df["Sales"],
         marker="o",color="darkblue",
         linewidth=2,label="Sales",
         markeredgecolor="black", # Marker edge color
         markeredgewidth=1.5, # Marker edge width
         markerfacecolor="skyblue") # Marker face color (fill)
plt.plot(sales_df["Month"],sales_df["Profit"],marker="s",linewidth=2,color="green",label="Profit",markeredgecolor="black",markeredgewidth=1.5,markerfacecolor="yellow") # Plot profit line with styling
plt.title("Monthly Sales and Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend(loc="upper left") # Display legend with location
plt.grid(True)
plt.show()
```

```python
labels = ["A", "B", "C"] # Define labels for pie slices
values = [50, 30, 20] # Define values for pie slices

plt.figure(figsize=(5,5))
plt.pie(values,labels=labels,autopct="%1.1f%%") # Create pie chart with labels and percentage format
plt.title("Pie Chart Example")
plt.show()
```

```python
products = ["Office Supplies", "Furniture", "Technology"]
sales = [719, 742, 836]
colours=["red","yellow","orange"] # Define colors for bars

plt.figure(figsize=(6,6))
plt.bar(products,sales,color=colours,edgecolor="black") # Create bar chart with colors and edge color
plt.xticks(rotation=45) # Rotate x-axis labels
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [1200, 1500, 1700, 1600]
profit = [200, 250, 300, 280]

fig,ax=plt.subplots(1,2,figsize=(10,6)) # Create figure and a grid of 1 row, 2 columns of subplots
ax[0].plot(months,sales,marker="o",color="violet",markerfacecolor="black") # Plot sales on first subplot
ax[0].set_title("Monthly Sales") # Set title for first subplot
ax[1].plot(months,profit,marker="s",color="pink",markerfacecolor="black") # Plot profit on second subplot
ax[1].set_title("Monthly Profit") # Set title for second subplot
plt.tight_layout() # Adjust layout to prevent overlaps

plt.savefig("Montly Profit and Sale Sample.png",dpi=300,bbox_inches="tight") # Save the figure to a file
```
