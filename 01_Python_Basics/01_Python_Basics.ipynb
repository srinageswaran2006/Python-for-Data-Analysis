## **Phase 3: Operators**

### Arithmetic Operators

```python
# Arithmetic Operators
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
```

### Comparison Operators

```python
# Comparison Operators
x = 20
y = 15

print(x > y)
print(x < y)
print(x == y)
print(x != y)
print(x >= y)
print(x <= y)
```

### Logical Operators

```python
# Logical Operators
age = 19
has_id = True

print(age >= 18 and has_id)
print(age < 18 or has_id)
print(not has_id)
```

### Assignment Operators

```python
# Assignment Operators
num = 10

num += 5
print(num)

num -= 3
print(num)

num *= 2
print(num)

num /= 4
print(num)
```

## **Phase 4: Conditional Statements**

### if Statement

```python
# if statement(basic)
age = 20
if age >= 18:
    print("You are eligible to vote.")
```

### if-else Statement

```python
# if-else statement
marks = 45
if marks >= 50:
    print("Pass")
else:
    print("Fail")
```

### if-elif-else statement

```python
# if-elif-else statement
marks = 78
if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Grade D")
```

### Nested if Statement

```python
# Nested if statement
age = 22
has_id = True
if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID card required")
else:
    print("Underage")
```

### Logical Operators in Condition

```python
# Using logical operators in conditions
username = "ram"
password = "python123"

if username == "ram" and password == "python123":
    print("Login successful")
else:
    print("Invalid credentials")
```

## **Phase 5: Loops**

### For Loop

```python
# Simple for loop(print 1-5)
for i in range(1, 6):
    print(i)
```

### For Loop with a List

```python
# Looping a list
fruits = ["apple", "banana", "mango"]
for fruit in fruits:
    print(fruit)
```

### While Loop

```python
#Simple while loop(print 1-5)
count = 1
while count <= 5:
    print(count)
    count += 1

#Infinite loop

# while True:
    # print("This is an infinite loop")
```

### Break Statement

```python
# loop breaks at i=4
for i in range(1, 10):
    if i == 4:
        break
    print(i)
```

### Continue Statement

```python
# Skips number 3
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

### Loop with else

```python
#using else in loop
for i in range(1, 4):
    print(i)
else:
    print("Loop completed")
```

### Nested Loops

```python
# Multiplication pattern
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
    print("---")
```

## **Phase 6: Functions**

### Creating a Function

```python
def greet(): #func definition
    print("Hello, Python!")
greet()
```

### Function with Parameters

```python
def greet(name): # name is parameter
    print(f"Hello, {name}!")

greet("Ram")
greet("Alex")
```

### Function with Return Value

```python
def add(a, b):
    return a + b #this value can be used all over in a program when returned
result = add(10, 5)
print("Sum:", result)
```

### Multiple Parameter

```python
def student_info(name, age, course): #multiple parameters
    return f"{name} is {age} years old and studies {course}."
print(student_info("Ram", 20, "Data Analysis"))
```

### Default Parameters

```python
def power(base, exponent=2): #default parameter is 2
    return base ** exponent

print(power(5))
print(power(5, 3))
```

### Keyword Arguments

```python
def employee(name, salary):
    print(f"{name} earns {salary}")

employee(salary=50000, name="Ram") #can pass values in any order when mentioned here
```

### Function in a Loop

```python
def square(num):
    return num ** 2

for i in range(1, 6):
    print(square(i)) #looping the funtion call
```

## **Phase 12: List Comprehensions**

### 12.1 Basic List Comprehension

```python
# Create a list of squares
squares=[x**2 for x in range(1,6)]
print(squares)
```

### 12.2 Using a Condition

```python
# Even numbers from 1 to 10
even_numbers=[x for x in range(1,11) if x%2==0]
print(even_numbers)
```

### 12.3 Odd Numbers

```python
# Odd numbers
odd_numbers=[x for x in range(1,11) if x%2!=0]
print(odd_numbers)
```

### 12.4 Convert Strings to Uppercase

```python
# Convert to uppercase
fruits=["apple","banana","mango"]
upper_fruits=[fruit.upper() for fruit in fruits]
print(upper_fruits)
```

### 12.5 Extract First Letters

```python
# First letter of each word
words=["Python","Data","Analysis"]
first_letters=[word[0] for word in words]
print(first_letters)
```

### 12.6 Replace Values Conditionally

```python
# Replace negative values with 0
numbers=[10,-5,20,-2,30]
cleaned=[x if x>=0 else 0 for x in numbers]
print(cleaned)
```

### 12.7 Create a List from a String

```python
# Extract vowels
text="dataanalysis"
vowels=[char for char in text if char in "aeiou"]
print(vowels)
```

### 12.8 Nested List Comprehension

```python
# Multiplication pairs
pairs=[(i,j) for i in range(1,4) for j in range(1,4)]
print(pairs)
```

### 12.9 Compare with a Normal Loop

```python
# Normal loop
numbers=[]
for i in range(1,6):
    numbers.append(i**2)
print(numbers)

# List comprehension
numbers=[i**2 for i in range(1,6)]
print(numbers)
```

### 13.9 File Modes: r+ and w+

```python
# r+ mode
#File must already exist
#Allows reading and writing
with open("sample.txt","r+") as file:
    content=file.read()
    print(content)
    file.write("\nAdded using r+ mode.")

# w+ mode
#Creates the file if it does not exist
#Deletes all existing content before writing
with open("sample2.txt","w+") as file:
    file.write("This file was created using w+ mode.")
    file.seek(0)
    print(file.read())
```

## **Phase 14: Exception Handling**

### 14.1 Basic try-except

```python
# Basic try-except
try:
    num=int(input("Enter a number: "))
    print("You entered:",num)
except:
    print("Invalid input!")
```

### 14.2 Handling a Specific Error

```python
# Handle ValueError
try:
    age=int(input("Enter your age: "))
    print(f"Your age is {age}.")
except ValueError:
    print("Please enter a valid integer.")
```

### 14.3 Handling Division by Zero

```python
# Handle ZeroDivisionError
try:
    a=int(input("Enter numerator: "))
    b=int(input("Enter denominator: "))
    print(a/b)
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

### 14.4 Multiple except Blocks

```python
# Multiple exceptions
try:
    a=int(input("Enter numerator: "))
    b=int(input("Enter denominator: "))
    print(a/b)
except ValueError:
    print("Enter valid integers.")
except ZeroDivisionError:
    print("Denominator cannot be zero.")
```

### 14.5 Using else

```python
# else block
try:
    num=int(input("Enter a number: "))
except ValueError:
    print("Invalid number.")
else:
    print(f"Square: {num**2}")
```

### 14.6 Using finally

```python
# finally block
try:
    num=int(input("Enter a number: "))
    print(num)
except ValueError:
    print("Invalid input.")
finally:
    print("Program finished.")
```

### 14.7 Raising Your Own Exception

```python
# raise an exception
#Intentionlly showing error
age=int(input("Enter age: "))
if age<18:
    raise ValueError("Age must be at least 18.")
print("Eligible")
```

### 14.8 Exception Handling with Files (Not-Workable)

```python
# Handle missing file
try:
    with open("missing.txt","r") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found.")
```
